### Target Distribution Analysis

This dataset contains three target classes, Graduate, Dropout and Enrolled.
Graduate has the highest number of student 2209 while Enrolled has the lowest number of student with 794.
The classes are unequally represented which indicates class imbalance.
So that accuracy checking will not be enough to evaluate models. We have to do other metrics such as Precision, Recall and F1 score should also be considered.

```python
feature_cols = ['Age at enrollment',
                'Admission grade',
                'Previous qualification (grade)',
                'Curricular units 1st sem (grade)',
                'Curricular units 2nd sem (grade)',
                'Unemployment rate',
                'Inflation rate',
                'GDP']

X = df[feature_cols]
y = df.Target
```

```python
X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.25, random_state=42
)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```
